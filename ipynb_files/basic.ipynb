# Olist E-commerce Dataset Inventory

This notebook establishes the analytical foundation for investigating delivery performance and customer experience in the Olist marketplace data. The first step is to verify which source tables are available, how large they are, and which business entities each table represents before designing joins or metrics.

```python
import pandas as pd
import numpy as np
```

## Source-table discovery

A lightweight inventory of the CSV files provides the schema and row-count baseline for the investigation. This is preferable to loading every table immediately: it confirms the available analytical grain and highlights the datasets that will require careful joins, such as orders, order items, payments, reviews, products, customers, and sellers.

```python
import pandas as pd, glob, os

files = sorted(glob.glob("../data/*.csv"))
print("Files found:", len(files))

for f in files:
    df = pd.read_csv(f, nrows=5)
    total = sum(1 for _ in open(f, encoding="utf-8")) - 1
    print(f"\n{os.path.basename(f)}  ({total} rows)")
    print(list(df.columns))
```

## Inventory findings and analytical implications

The inventory contains 9 source files spanning 1,000,163 geolocation records, 112,650 order-item records, 104,719 reviews, and 99,441 orders and customers. The order table exposes purchase, approval, carrier-delivery, customer-delivery, and estimated-delivery timestamps, making it the central source for measuring delivery timeliness; item, payment, review, product, customer, seller, and geolocation tables provide the dimensions needed to compare operational and customer outcomes.

Because the tables are recorded at different grains, subsequent analysis should join them through their documented identifiers and aggregate before making order-level comparisons. At this stage the notebook establishes data coverage and schema readiness; no delivery-risk rate, model result, or causal business conclusion has yet been computed.
